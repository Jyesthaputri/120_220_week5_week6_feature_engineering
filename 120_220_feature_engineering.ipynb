```python
import pandas as pd
import re

df = pd.read_csv('reelscomment_preprocessed_fix.csv')

print("Ukuran dataset:", df.shape)
print("Nama kolom:", df.columns.tolist())
print("\nContoh komentar:")
display(df[['9_stemming']].head())
```

```python
df_pre = df['9_stemming'].dropna()

print("Jumlah dokumen:", len(df_pre))
print("\nContoh teks hasil preprocessing:")
display(df_pre.head())
```

One-Hot Encoding

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer_binary = CountVectorizer(binary=True)
X_binary = vectorizer_binary.fit_transform(df_pre)

df_binary = pd.DataFrame(
    X_binary.toarray(),
    columns=vectorizer_binary.get_feature_names_out()
)

print("Ukuran matriks:", df_binary.shape)
display(df_binary.head(10))
```

```python
vectorizer_count = CountVectorizer()
X_count = vectorizer_count.fit_transform(df_pre)

df_count = pd.DataFrame(
    X_count.toarray(),
    columns=vectorizer_count.get_feature_names_out()
)

print("Ukuran matriks:", df_count.shape)
display(df_count.head())
```

TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer_tfidf = TfidfVectorizer()
X_tfidf = vectorizer_tfidf.fit_transform(df_pre)

df_tfidf = pd.DataFrame(
    X_tfidf.toarray(),
    columns=vectorizer_tfidf.get_feature_names_out()
)

print("Ukuran matriks:", df_tfidf.shape)
display(df_tfidf.head())
```

```python
#Tokenize untuk Word2Vec dan FastText

import nltk
from nltk import word_tokenize
from tqdm.auto import tqdm

nltk.download('punkt')
nltk.download('punkt_tab')

sentences = [
    word_tokenize(comment.lower())
    for comment in tqdm(df_pre, desc="Tokenisasi")
]

print("Contoh token:")
print(sentences[:5])
```

Word2Vec

```python
!pip install -q gensim
```

```python
import gensim.models

# 1. Train model
model = gensim.models.Word2Vec(
    sentences,
    min_count=1,
    vector_size=100,
    window=5,
    sg=0,
    seed=42,
    workers=1
)

# 2. Save model
model.save('sentimen_reels.w2v')

# 3. Load model
model = gensim.models.Word2Vec.load('sentimen_reels.w2v')

# 4. Model information
w2v = model.wv

print("10 kata pertama:", w2v.index_to_key[:10])
print("Ukuran matriks dan vektor:", w2v.vectors.shape)
print("Dimensi vektor:", w2v.vector_size)
```

```python
kata = 'pakai'

if kata in w2v:
    print(w2v.similar_by_key(kata, topn=5))
else:
    print(f"Kata '{kata}' tidak ditemukan dalam kosakata.")

    if len(w2v.index_to_key) > 0:
        kata = w2v.index_to_key[0]
        print("Contoh kata alternatif:", kata)
        print(w2v.similar_by_key(kata, topn=5))
```

```python
import numpy as np

def get_document_vector(tokens, model):
    valid_vectors = [
        model.wv[word]
        for word in tokens
        if word in model.wv
    ]

    if not valid_vectors:
        return np.zeros(model.vector_size)

    return np.mean(valid_vectors, axis=0)

X_features_w2v = np.array([
    get_document_vector(doc, model)
    for doc in sentences
])

df_w2v = pd.DataFrame(X_features_w2v)

print("Ukuran matriks fitur dokumen:", df_w2v.shape)
display(df_w2v.head())
```

Visualisasi Word2Vec

```python
!pip install -q umap-learn plotly
```

```python
from umap import UMAP
import plotly.express as px

if len(w2v.index_to_key) >= 3:
    reducer = UMAP(
        n_components=2,
        random_state=42,
        n_neighbors=min(30, len(w2v.index_to_key) - 1)
    )

    X_umap_w2v = reducer.fit_transform(w2v.vectors)

    df_umap_w2v = pd.DataFrame(
        X_umap_w2v,
        columns=['umap1', 'umap2']
    )

    df_umap_w2v['word'] = w2v.index_to_key

    fig = px.scatter(
        df_umap_w2v,
        x='umap1',
        y='umap2',
        text='word',
        title='Word2Vec Visualization'
    )

    fig.update_traces(textposition='top center')
    fig.update_layout(height=800)
    fig.show()
else:
    print("Kosakata terlalu sedikit untuk visualisasi UMAP.")
```

FastText

```python
import gensim.models

# 1. Build model
model_fasttext = gensim.models.FastText(
    sentences,
    min_count=2,
    vector_size=100,
    window=5,
    min_n=3,
    max_n=6,
    sg=1,
    epochs=10,
    seed=42,
    workers=1
)

# 2. Save model
model_fasttext.save('sentimen_reels.fasttext')

# 3. Load model
model_fasttext = gensim.models.FastText.load(
    'sentimen_reels.fasttext'
)

# 4. Model information
fastext = model_fasttext.wv

print("10 kata pertama:", fastext.index_to_key[:10])
print("Ukuran matriks vektor:", fastext.vectors.shape)
print("Dimensi vektor:", fastext.vector_size)
```

```python
kata = '4g'

if kata in fastext:
    print(fastext.similar_by_key(kata, topn=5))
else:
    print(f"Kata '{kata}' tidak ditemukan dalam kosakata.")

    if len(fastext.index_to_key) > 0:
        kata = fastext.index_to_key[0]
        print("Contoh kata alternatif:", kata)
        print(fastext.similar_by_key(kata, topn=5))
```

```python
def get_document_vector_fasttext(tokens, model):
    valid_vectors = [
        model.wv[word]
        for word in tokens
        if word in model.wv
    ]

    if not valid_vectors:
        return np.zeros(model.vector_size)

    return np.mean(valid_vectors, axis=0)

X_features_fasttext = np.array([
    get_document_vector_fasttext(doc, model_fasttext)
    for doc in sentences
])

df_fasttext = pd.DataFrame(X_features_fasttext)

print("Ukuran matriks fitur dokumen:", df_fasttext.shape)
display(df_fasttext.head())
```

Visualisasi FastText

```python
if len(fastext.index_to_key) >= 3:
    reducer_fasttext = UMAP(
        n_components=2,
        random_state=42,
        min_dist=0.3,
        n_neighbors=min(25, len(fastext.index_to_key) - 1)
    )

    X_umap_fasttext = reducer_fasttext.fit_transform(
        fastext.vectors
    )

    df_umap_fasttext = pd.DataFrame(
        X_umap_fasttext,
        columns=['umap1', 'umap2']
    )

    df_umap_fasttext['word'] = fastext.index_to_key

    fig = px.scatter(
        df_umap_fasttext,
        x='umap1',
        y='umap2',
        text='word',
        title='FastText Visualization'
    )

    fig.update_traces(textposition='top center')
    fig.update_layout(height=800)
    fig.show()
else:
    print("Kosakata terlalu sedikit untuk visualisasi UMAP.")
```
